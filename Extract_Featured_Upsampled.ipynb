```python
# POOL 3 BUT WE NEED POOL 4

"""
UPSAMPLED DATASET
VGG-16 Feature Extractor for Old Greek Letters
Cuts VGG-16 at block3_pool → GlobalAveragePooling → 256-d vector per image.
Outputs: upsampled_features_block3pool.npy, upsampled_labels_block3pool.npy, upsampled_class_indices.json
"""

import json
import os
import numpy as np
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator


# ── Settings ───────────────────────────────────────────────────────────────────

DATA_DIR   = "C:/Users/niric/Desktop/classes/AI and ML/CLEAN_DATASET_UPSAMPLED"
OUTPUT_DIR = "C:/Users/niric/Desktop/classes/AI and ML/UPSAMPLED_OUTPUT"     # folder where .npy and .json files will be saved
IMG_SIZE   = (70, 70)
BATCH_SIZE = 64


# ── Step 1: Build the feature extractor ───────────────────────────────────────

base_model = VGG16(weights="imagenet", include_top=False, input_shape=(*IMG_SIZE, 3))

for layer in base_model.layers:
    layer.trainable = False

# Grab the output at block3_pool (shape 8×8×256) — everything after is ignored
cut_output = base_model.get_layer("block3_pool").output

# Average each of the 256 channels across the 8×8 grid → flat 256-d vector
pooled = GlobalAveragePooling2D()(cut_output)

model = Model(inputs=base_model.input, outputs=pooled)


# ── Step 2: Load images ────────────────────────────────────────────────────────

datagen   = ImageDataGenerator(preprocessing_function=preprocess_input)
generator = datagen.flow_from_directory(
    DATA_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="sparse",   # labels as integers: Alpha=0, Beta=1, ...
    shuffle=False,         # must be False so features and labels stay aligned
    color_mode="rgb"
)

total_batches = int(np.ceil(generator.samples / BATCH_SIZE))
print(f"Found {generator.samples} images across {generator.num_classes} classes")


# ── Step 3: Extract features ───────────────────────────────────────────────────

all_features = []
all_labels   = []

for batch_number, (batch_images, batch_labels) in enumerate(generator):

    all_features.append(model.predict(batch_images, verbose=0))
    all_labels.append(batch_labels)

    print(f"  Batch {batch_number + 1} / {total_batches}")

    if batch_number + 1 >= total_batches:
        break

all_features = np.vstack(all_features)                       # shape: (N, 256)
all_labels   = np.concatenate(all_labels).astype(np.int32)  # shape: (N,)

print(f"Features: {all_features.shape},  Labels: {all_labels.shape}")


# ── Step 4: Save ───────────────────────────────────────────────────────────────

os.makedirs(OUTPUT_DIR, exist_ok=True)   # creates the folder if it doesn't exist yet

np.save(f"{OUTPUT_DIR}/upsampled_features_block3pool.npy", all_features)
np.save(f"{OUTPUT_DIR}/upsampled_labels_block3pool.npy",   all_labels)

with open(f"{OUTPUT_DIR}/upsampled_class_indices.json", "w") as f:
    json.dump(generator.class_indices, f, indent=2)  # e.g. {"Alpha": 0, "Beta": 1, ...}

print(f"Saved all files to {OUTPUT_DIR}/")
```
