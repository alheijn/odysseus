```python
import os
import sys
from pathlib import Path

try:
    import kaggle
except OSError:
    # The Kaggle API requires authentication upon import. 
    print("Kaggle API credentials not found!")
    print("To authenticate, please set the KAGGLE_API_TOKEN environment variable:")
    print("  export KAGGLE_API_TOKEN=KGAT_xxxxx")
    print("Alternatively, save your token to a file named 'access_token' in ~/.kaggle/ (Mac/Linux) or C:\\Users\\<User>\\.kaggle\\ (Windows).")
    sys.exit(1)

# The exact Kaggle dataset identifier from the URL
DATASET_IDENTIFIER = "miswindall/al-pub-v2"

# The target directory defined in the Project Odysseus README
TARGET_DIR = Path("../data")

def download_al_pub_v2():
    """
    Downloads and extracts the AL-PUB_v2 dataset into the target directory.
    """
    print("=" * 50)
    print("Project Odysseus: AL-PUB_v2 Dataset Downloader")
    print("=" * 50)
    
    # Create the data directory if it doesn't exist yet
    TARGET_DIR.mkdir(parents=True, exist_ok=True)
    print(f"Target directory set to: {TARGET_DIR.absolute()}")
    
    print(f"Downloading dataset '{DATASET_IDENTIFIER}' from Kaggle...")
    print("This may take a moment depending on your internet connection.")
    
    # The kaggle API handles downloading and unzipping in one command
    try:
        kaggle.api.dataset_download_files(
            DATASET_IDENTIFIER, 
            path=str(TARGET_DIR), 
            unzip=True,      # Automatically extracts the contents
            quiet=False      # Shows a progress bar
        )
        print("Download and extraction complete!")
        print(f"The Ancient Greek character images are now available in the '{TARGET_DIR}' folder.")
        
    except Exception as e:
        print(f"\nAn error occurred during the download process:\n{e}")

if __name__ == "__main__":
    download_al_pub_v2()
```
